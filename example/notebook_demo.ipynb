# kgviz notebook demo

Interactive 3D knowledge graph in Jupyter.

```bash
pip install kgviz ipython
# optional: pip install "kgviz[maps]"
```

Run all cells. The graph renders inline via `_repr_html_()`.

````python
from kgviz import Graph3D

nodes = [
    {"id": "A", "label": "Alpha", "color": "#ff6b6b", "size": 8},
    {"id": "B", "label": "Beta", "color": "#00cc88", "size": 6},
    {"id": "C", "label": "Gamma", "color": "#45b7d1", "size": 5},
    {"id": "D", "label": "Delta", "color": "#f0b429", "size": 4},
]
edges = [
    {"source": "A", "target": "B"},
    {"source": "B", "target": "C"},
    {"source": "C", "target": "D"},
    {"source": "A", "target": "D"},
]

fig = Graph3D(nodes=nodes, edges=edges, show_labels=True, height=520)
fig
````

## Embedding map (optional)

Requires `pip install "kgviz[maps]"`.

````python
import numpy as np

topics = ["ML", "Biology", "Physics", "Chemistry"]
map_nodes = [
    {"id": i, "label": f"Paper {i}", "topic": topics[i % len(topics)]}
    for i in range(24)
]
features = np.random.default_rng(42).standard_normal((24, 16))

map_fig = Graph3D.from_map(map_nodes, features, method="pca", node_color_by="topic")
map_fig
````

## Export HTML

Open `notebook_export.html` in any browser.

````python
from pathlib import Path

out = Path("notebook_export.html")
out.write_text(fig.to_html(), encoding="utf-8")
print(out.resolve())
````
