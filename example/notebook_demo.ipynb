# kgviz notebook demo

Interactive 3D knowledge graph in Jupyter.

```bash
pip install kgviz ipython
# optional: pip install "kgviz[maps]"
```

Run all cells. The graph renders inline via `_repr_html_()`.

````python
from kgviz import Graph3D

nodes = [
    {"id": "A", "label": "Alpha", "color": "#ff6b6b", "size": 8},
    {"id": "B", "label": "Beta", "color": "#00cc88", "size": 6},
    {"id": "C", "label": "Gamma", "color": "#45b7d1", "size": 5},
    {"id": "D", "label": "Delta", "color": "#f0b429", "size": 4},
]
edges = [
    {"source": "A", "target": "B"},
    {"source": "B", "target": "C"},
    {"source": "C", "target": "D"},
    {"source": "A", "target": "D"},
]

fig = Graph3D(nodes=nodes, edges=edges, show_labels=True, height=520)
fig
````

## Embedding map (optional)

Same clustered demo data as [`demo_map_tsne.html`](https://github.com/dataprofessor/kgviz/blob/main/example/demo_map_tsne.html). Requires `pip install "kgviz[maps]"`.

````python
import sys
from pathlib import Path

from kgviz import Graph3D

# demo_graph.py lives next to this notebook (Colab clone or local repo)
_example = Path.cwd() / "example" if (Path.cwd() / "example" / "demo_graph.py").is_file() else Path.cwd()
if str(_example) not in sys.path:
    sys.path.insert(0, str(_example.resolve()))

from demo_graph import synthetic_corpus

map_nodes, features = synthetic_corpus(400)
map_fig = Graph3D.from_map(
    map_nodes,
    features,
    method="tsne",
    knn_k=0,
    node_color_by="topic",
    legend_node_by="topic",
    height=700,
)
map_fig
````
